```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from IPython.display import display
from src.database import get_connection

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
pd.set_option("display.float_format", "{:,.2f}".format)

sns.set_theme(style="whitegrid")

print("Libraries imported successfully.")
```

### Connect to SQL Server

```python
conn = get_connection()

print("Connected to DecathlonDB successfully.")
```

### Check available tables

```python
tables_query = """
SELECT
    TABLE_SCHEMA,
    TABLE_NAME
FROM INFORMATION_SCHEMA.TABLES
WHERE TABLE_TYPE = 'BASE TABLE'
ORDER BY TABLE_SCHEMA, TABLE_NAME;
"""

tables_df = pd.read_sql_query(tables_query, conn)

display(tables_df)
```

### Load the three tables

```python
products_query = "SELECT * FROM dbo.products"
sales_query = "SELECT * FROM dbo.sales"
store_query = "SELECT * FROM dbo.store"

products_df = pd.read_sql_query(products_query, conn)
sales_df = pd.read_sql_query(sales_query, conn)
store_df = pd.read_sql_query(store_query, conn)

print("All tables loaded successfully.")

print(f"Products: {products_df.shape[0]:,} rows")
print(f"Sales:    {sales_df.shape[0]:,} rows")
print(f"Stores:   {store_df.shape[0]:,} rows")
```

### Preview the datasets

```python
print("PRODUCTS DATA")
display(products_df.head())

print("SALES DATA")
display(sales_df.head())

print("STORE DATA")
display(store_df.head())
```

### Understand dataset dimensions

```python
dataset_summary = pd.DataFrame({
    "Dataset": ["Products", "Sales", "Stores"],
    "Rows": [
        products_df.shape[0],
        sales_df.shape[0],
        store_df.shape[0]
    ],
    "Columns": [
        products_df.shape[1],
        sales_df.shape[1],
        store_df.shape[1]
    ]
})

display(dataset_summary)
```

### Inspect column names

```python
datasets = {
    "Products": products_df,
    "Sales": sales_df,
    "Stores": store_df
}

for name, df in datasets.items():
    print(f"\n{name.upper()} COLUMN NAMES")
    print(df.columns.tolist())
```

### Inspect data types

```python
for name, df in datasets.items():
    print(f"\n{name.upper()} DATA TYPES")
    df.info()
```

### Create a data type summary

```python
for name, df in datasets.items():
    dtype_summary = pd.DataFrame({
        "Column": df.columns,
        "Data_Type": df.dtypes.astype(str).values,
        "Non_Null_Count": df.notna().sum().values,
        "Missing_Count": df.isna().sum().values
    })

    print(f"\n{name.upper()} DATA TYPE SUMMARY")
    display(dtype_summary)
```

### Check missing values

```python
for name, df in datasets.items():
    print(f"\n{name.upper()} MISSING VALUES")

    missing_summary = pd.DataFrame({
        "Missing_Count": df.isna().sum(),
        "Missing_Percentage": (
            df.isna().sum() / len(df) * 100
            if len(df) > 0
            else 0
        )
    })

    missing_summary = missing_summary.sort_values(
        by="Missing_Percentage",
        ascending=False
    )

    display(missing_summary)
```

### Visualize missing values

```python
for name, df in datasets.items():
    missing_counts = df.isna().sum()
    missing_counts = missing_counts[missing_counts > 0]

    if not missing_counts.empty:
        plt.figure(figsize=(10, 5))
        missing_counts.sort_values().plot(kind="barh")

        plt.title(f"Missing Values — {name}")
        plt.xlabel("Number of Missing Values")
        plt.ylabel("Columns")
        plt.tight_layout()
        plt.show()
    else:
        print(f"{name}: No missing values found.")
```

### Check duplicate records

```python
for name, df in datasets.items():
    duplicate_count = df.duplicated().sum()

    print(f"{name}: {duplicate_count:,} duplicate rows")
```

### Display duplicate records

```python
for name, df in datasets.items():
    duplicates = df[df.duplicated(keep=False)]

    print(f"\n{name.upper()} DUPLICATE RECORDS")

    if duplicates.empty:
        print("No duplicate rows found.")
    else:
        display(duplicates.head(10))
```

### Generate descriptive statistics

```python
for name, df in datasets.items():
    print(f"\n{name.upper()} NUMERICAL SUMMARY")

    display(df.describe().T)
```

### Explore categorical columns

```python
for name, df in datasets.items():
    categorical_columns = df.select_dtypes(
        include=["object", "category", "string"]
    ).columns

    print(f"\n{name.upper()} CATEGORICAL COLUMNS")

    for column in categorical_columns:
        print(f"\nColumn: {column}")
        print(f"Unique values: {df[column].nunique(dropna=True)}")

        display(
            df[column]
            .value_counts(dropna=False)
            .head(10)
            .rename_axis(column)
            .reset_index(name="Count")
        )
```

### Check unique values in identifier columns
This step helps you inspect identifiers without assuming their names in advance.

```python
for name, df in datasets.items():
    print(f"\n{name.upper()} IDENTIFIER COLUMNS")

    identifier_columns = [
        column for column in df.columns
        if any(
            keyword in column.lower()
            for keyword in ["id", "key", "code"]
        )
    ]

    if not identifier_columns:
        print("No obvious identifier columns found.")
        continue

    for column in identifier_columns:
        print(f"\nColumn: {column}")
        print(f"Unique values: {df[column].nunique(dropna=True)}")
        print(f"Missing values: {df[column].isna().sum()}")

        display(df[column].value_counts().head(10))
```

### Check potential negative values
Negative values may be valid in some business situations, such as refunds or sales adjustments. This cell only identifies them; it does not change the data.

```python
for name, df in datasets.items():
    numeric_columns = df.select_dtypes(
        include=np.number
    ).columns

    print(f"\n{name.upper()} NEGATIVE VALUE CHECK")

    found_negative = False

    for column in numeric_columns:
        negative_count = (df[column] < 0).sum()

        if negative_count > 0:
            found_negative = True
            print(
                f"{column}: {negative_count:,} negative values"
            )

    if not found_negative:
        print("No negative values found in numeric columns.")
```

### Check date columns
This cell detects columns whose names suggest they contain dates or timestamps. It does not permanently change the original DataFrames.

```python
for name, df in datasets.items():
    print(f"\n{name.upper()} DATE COLUMN CHECK")

    date_columns = [
        column for column in df.columns
        if any(
            keyword in column.lower()
            for keyword in ["date", "time"]
        )
    ]

    if not date_columns:
        print("No obvious date or time columns found.")
        continue

    for column in date_columns:
        converted_dates = pd.to_datetime(
            df[column],
            errors="coerce"
        )

        invalid_count = (
            df[column].notna() & converted_dates.isna()
        ).sum()

        print(f"\nColumn: {column}")
        print(f"Original data type: {df[column].dtype}")
        print(f"Missing values: {df[column].isna().sum()}")
        print(f"Unparseable non-missing values: {invalid_count}")

        valid_dates = converted_dates.dropna()

        if not valid_dates.empty:
            print(f"Earliest date: {valid_dates.min()}")
            print(f"Latest date: {valid_dates.max()}")
```

### Check likely business-value columns
For the Decathlon sales dataset, we should investigate prices, costs, quantities, discounts, and revenue if those columns exist.

```python
for name, df in datasets.items():
    print(f"\n{name.upper()} BUSINESS VALUE CHECK")

    keywords = [
        "price",
        "cost",
        "quantity",
        "discount",
        "revenue",
        "sales",
        "profit",
        "amount"
    ]

    candidate_columns = [
        column for column in df.columns
        if any(
            keyword in column.lower()
            for keyword in keywords
        )
    ]

    if not candidate_columns:
        print("No obvious business-value columns found.")
        continue

    for column in candidate_columns:
        print(f"\nColumn: {column}")
        print(f"Data type: {df[column].dtype}")
        print(f"Missing values: {df[column].isna().sum()}")

        numeric_values = pd.to_numeric(
            df[column],
            errors="coerce"
        )

        if numeric_values.notna().any():
            print(f"Minimum: {numeric_values.min()}")
            print(f"Maximum: {numeric_values.max()}")
```

### Explore sales by product, if the relevant columns exist

The following cell checks whether your sales and products tables share an obvious product identifier. It then uses the available column names to summarize sales by product.

```python
def find_matching_column(df, candidates):
    column_lookup = {
        column.lower(): column
        for column in df.columns
    }

    for candidate in candidates:
        if candidate.lower() in column_lookup:
            return column_lookup[candidate.lower()]

    return None


sales_product_key = find_matching_column(
    sales_df,
    ["Product_ID", "Product_Key", "ProductID", "ProductKey"]
)

products_product_key = find_matching_column(
    products_df,
    ["Product_ID", "Product_Key", "ProductID", "ProductKey"]
)

quantity_column = find_matching_column(
    sales_df,
    ["Quantity", "Units_Sold", "Units"]
)

revenue_column = find_matching_column(
    sales_df,
    ["Gross_Sales_INR", "Revenue", "Sales_Amount", "Total_Sales"]
)

product_name_column = find_matching_column(
    products_df,
    ["Product_Name", "ProductName", "Name"]
)

if sales_product_key and products_product_key:
    print("Potential product key found.")
    print("Sales key:", sales_product_key)
    print("Products key:", products_product_key)

    if quantity_column or revenue_column:
        sales_columns = [sales_product_key]

        if quantity_column:
            sales_columns.append(quantity_column)

        if revenue_column:
            sales_columns.append(revenue_column)

        product_sales = sales_df[sales_columns].copy()

        product_sales = product_sales.merge(
            products_df[
                [products_product_key]
                + ([product_name_column] if product_name_column else [])
            ].drop_duplicates(),
            left_on=sales_product_key,
            right_on=products_product_key,
            how="left",
            indicator=True
        )

        group_columns = [sales_product_key]

        if product_name_column:
            group_columns.append(product_name_column)

        aggregations = {}

        if quantity_column:
            aggregations[quantity_column] = "sum"

        if revenue_column:
            aggregations[revenue_column] = "sum"

        product_summary = (
            product_sales
            .groupby(group_columns, dropna=False)
            .agg(aggregations)
            .reset_index()
            .sort_values(
                by=list(aggregations.keys())[0],
                ascending=False
            )
        )

        display(product_summary.head(10))

        unmatched_count = (
            product_sales["_merge"] == "left_only"
        ).sum()

        print("Sales rows without a matching product:", unmatched_count)

    else:
        print(
            "A product key exists, but no recognized quantity "
            "or revenue column was found."
        )
else:
    print(
        "Could not identify matching product keys automatically. "
        "Review the column names and database relationships."
    )
```

### Check possible relationships between sales and stores

```python
sales_store_key = find_matching_column(
    sales_df,
    ["Store_ID", "Store_Key", "StoreID", "StoreKey"]
)

store_key = find_matching_column(
    store_df,
    ["Store_ID", "Store_Key", "StoreID", "StoreKey"]
)

if sales_store_key and store_key:
    print("Potential store relationship found.")
    print("Sales table key:", sales_store_key)
    print("Store table key:", store_key)

    store_keys = store_df[store_key].dropna().unique()

    unmatched_sales = ~sales_df[sales_store_key].isin(store_keys)

    print(
        "Sales rows with no matching store:",
        int(unmatched_sales.sum())
    )

    print(
        "Distinct store identifiers in sales:",
        sales_df[sales_store_key].nunique(dropna=True)
    )

    print(
        "Distinct store identifiers in store table:",
        store_df[store_key].nunique(dropna=True)
    )
else:
    print(
        "Could not identify matching store keys automatically. "
        "Review the column names and database relationships."
    )
```

### Build a data quality summary
This cell creates one summary table showing the overall data quality of each dataset.

```python
quality_results = []

for name, df in datasets.items():
    total_cells = df.shape[0] * df.shape[1]
    missing_cells = int(df.isna().sum().sum())

    quality_results.append({
        "Dataset": name,
        "Rows": df.shape[0],
        "Columns": df.shape[1],
        "Missing_Cells": missing_cells,
        "Missing_Percentage": (
            missing_cells / total_cells * 100
            if total_cells > 0
            else 0
        ),
        "Duplicate_Rows": int(df.duplicated().sum())
    })

quality_summary = pd.DataFrame(quality_results)

display(quality_summary)
```

```python
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent
REPORTS_DIR = PROJECT_ROOT / "reports"

REPORTS_DIR.mkdir(parents=True, exist_ok=True)

quality_summary.to_csv(
    REPORTS_DIR / "data_quality_summary.csv",
    index=False
)

for name, df in datasets.items():
    column_summary = pd.DataFrame({
        "Column": df.columns,
        "Data_Type": df.dtypes.astype(str).values,
        "Non_Null_Count": df.notna().sum().values,
        "Missing_Count": df.isna().sum().values,
        "Unique_Values": [
            df[column].nunique(dropna=True)
            for column in df.columns
        ]
    })

    file_name = name.lower() + "_column_summary.csv"

    column_summary.to_csv(
        REPORTS_DIR / file_name,
        index=False
    )

print("Exploration summaries saved to:", REPORTS_DIR)
```
