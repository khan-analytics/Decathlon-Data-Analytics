## 1. Database Connection

This notebook establishes a connection between Python and
Microsoft SQL Server.

**Database:** DecathlonDB

**Purpose:** Verify the database connection before performing
data exploration and business analysis.

### Import libraries

```python
import pyodbc
import pandas as pd
```

### Configure the connection

```python
import pyodbc

DRIVER_NAME = 'SQL Server'
SERVER_NAME = r'NASIRS-LAPTOP\SQLEXPRESS'
DATABASE_NAME = 'DecathlonDB'

connection_string = (
    f'DRIVER={{{DRIVER_NAME}}};'
    f'SERVER={SERVER_NAME};'
    f'DATABASE={DATABASE_NAME};'
    'Trusted_Connection=yes;'
)
```

### Establish the connection

```python
conn = pyodbc.connect(connection_string)

print("Connected successfully!")
```

### Verify the database

```python
cursor = conn.cursor()

cursor.execute("SELECT DB_NAME()")

print("Connected Database:", cursor.fetchone()[0])
```

### Verify the available tables

```python
query = """
SELECT
    TABLE_SCHEMA,
    TABLE_NAME
FROM INFORMATION_SCHEMA.TABLES
WHERE TABLE_TYPE = 'BASE TABLE'
ORDER BY TABLE_SCHEMA, TABLE_NAME;
"""

tables = pd.read_sql_query(query, conn)

tables
```
